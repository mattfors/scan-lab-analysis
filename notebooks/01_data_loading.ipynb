## Load and inspect the processed data

```python
import pandas as pd

# Load the processed parquet file
scans_df = pd.read_parquet('/workspaces/scan-lab-analysis/data/processed/scans.parquet')

print(f"Shape: {scans_df.shape}")
print(f"\nColumns: {list(scans_df.columns)}")
print(f"\nData types:\n{scans_df.dtypes}")
print(f"\nFirst few rows:")
print(scans_df.head(10))
```

## Summary statistics

```python
print(f"Number of experiments: {scans_df['experimentId'].nunique()}")
print(f"Number of users: {scans_df['userName'].nunique()}")
print(f"Number of unique barcodes: {scans_df['barcode'].nunique()}")
print(f"\nScan style distribution:")
print(scans_df['targetScanStyle'].value_counts())
print(f"\nTarget cluster size distribution:")
print(scans_df['targetClusterSize'].value_counts())
```

## Scans per experiment

```python
scans_per_experiment = scans_df.groupby('experimentId').size()
print(f"Scans per experiment:")
print(f"  Min: {scans_per_experiment.min()}")
print(f"  Max: {scans_per_experiment.max()}")
print(f"  Mean: {scans_per_experiment.mean():.1f}")
print(f"\nDistribution:")
print(scans_per_experiment.describe())
```
